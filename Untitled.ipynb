```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mutual_info_score
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('course_lead_scoring_2026.csv')
df.head()
```

```python
df['industry'].value_counts()
```

```python
numerical = ['annual_income', 'number_of_courses_viewed', 'interaction_count', 'lead_score']
df[numerical].corr()
```

```python
categorical = ['lead_source', 'industry', 'employment_status', 'location']
numerical = ['annual_income', 'number_of_courses_viewed', 'interaction_count', 'lead_score']

for c in categorical:
    df[c] = df[c].fillna('NA')
for c in numerical:
    df[c] = df[c].fillna(0.0)
```

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=42)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=42)

df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

y_train = df_train.converted.values
y_val = df_val.converted.values
y_test = df_test.converted.values

del df_train['converted']
del df_val['converted']
del df_test['converted']
```

```python
def mutual_info_converted_score(series):
    return mutual_info_score(series, df_full_train.converted)

mi = df_full_train[categorical].apply(mutual_info_converted_score)
mi.round(2).sort_values(ascending=False)
```

```python
dv = DictVectorizer(sparse=False)

train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)

val_dict = df_val[categorical + numerical].to_dict(orient='records')
X_val = dv.transform(val_dict)

model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict_proba(X_val)[:, 1]
churn_decision = (y_pred >= 0.5)
accuracy = (y_val == churn_decision).mean()
print(round(accuracy, 2))
```

```python
base_accuracy = accuracy

for feature in ['lead_source', 'number_of_courses_viewed', 'interaction_count']:
    features_without = [f for f in categorical + numerical if f != feature]
    dicts_train_wo = df_train[features_without].to_dict(orient='records')
    dicts_val_wo = df_val[features_without].to_dict(orient='records')

    dv_wo = DictVectorizer(sparse=False)
    X_train_wo = dv_wo.fit_transform(dicts_train_wo)
    X_val_wo = dv_wo.transform(dicts_val_wo)

    model_wo = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
    model_wo.fit(X_train_wo, y_train)

    y_pred_wo = model_wo.predict_proba(X_val_wo)[:, 1]
    decision_wo = (y_pred_wo >= 0.5)
    accuracy_wo = (y_val == decision_wo).mean()

    diff = base_accuracy - accuracy_wo
    print(feature, round(diff, 4))
```

```python
for C in [0.000001, 0.00001, 0.0001, 0.001]:
    model_c = LogisticRegression(solver='liblinear', C=C, max_iter=1000, random_state=42)
    model_c.fit(X_train, y_train)

    y_pred_c = model_c.predict_proba(X_val)[:, 1]
    decision_c = (y_pred_c >= 0.5)
    accuracy_c = (y_val == decision_c).mean()

    print(C, round(accuracy_c, 3))
```
